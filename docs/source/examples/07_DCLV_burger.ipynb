## burger dataset

```python
# poison dataset
from clustvartools.datasets import burger
data = burger.data
data.info()
```

## Instanciation and training

```python
# instanciation
from clustvartools import DCLV
clf = DCLV(sup_var=(10,11),max_iter=0)
```

### `fit` function

```python
# fit function
clf.fit(data)
```

### `fit_transform` function

```python
# fit_transform function - squared loadings
sqload = clf.fit_transform(data)
sqload
```

### `transform` function

```python
# transform function
sqload = clf.transform(burger.actif)
sqload
```

### `fit_predict` function

```python
# fit_predict function
prediction = clf.fit_predict(data)
prediction.to_frame()
```

### `predict` function

```python
# predict function
prediction = clf.predict(burger.actif)
prediction.to_frame()
```
